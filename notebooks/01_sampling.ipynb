```python
import pandas as pd
import numpy as np
from tqdm import tqdm
```

```python
def sample_large_csv(path, sample_frac=0.15):
    df = pd.read_parquet(path)
    sampled_df = df.sample(frac=sample_frac, random_state=42)
    return sampled_df
```

```python
import os
print(os.getcwd())
```

```python
base_path = "../data/raw/cicddos2019/"

syn_sample = sample_large_csv(base_path + "Syn-testing.parquet", sample_frac=0.15)
udp_sample = sample_large_csv(base_path + "UDP-training.parquet", sample_frac=0.15)
udplag_sample = sample_large_csv(base_path + "UDPLag-training.parquet", sample_frac=0.15)
```

```python
combined = pd.concat([syn_sample, udp_sample, udplag_sample], ignore_index=True)
combined.to_csv("../data/processed/cic_sampled.csv", index=False)
```

```python
drop_cols = [
    "Flow ID",
    "Source IP",
    "Destination IP",
    "Timestamp",
    "SimillarHTTP",  # typo in dataset
    "Unnamed: 0"
]

for col in drop_cols:
    if col in final.columns:
        final.drop(columns=[col], inplace=True)
```

```python
for col in final.columns:
    if final[col].dtype == 'object':
        final[col] = pd.to_numeric(final[col], errors='coerce')
```

```python
final.replace([np.inf, -np.inf], np.nan, inplace=True)
final.fillna(0, inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
final['Label_encoded'] = le.fit_transform(final['Label'])
le.classes_
```
